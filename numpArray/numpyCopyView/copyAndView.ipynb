```python
import numpy as np

arr = np.array([34,90,45,78,55,23,88,57])
### Copy method :: Original Array not affected
copy_arr = arr.copy()
copy_arr[1] = 89
copy_arr[-1] = 66
print("Original Array " ,arr)
print("Copied Array " ,copy_arr)
```

```python
import numpy as np

arr = np.array([34,90,45,78,55,23,88,57])
### View method :: Original Array  affected
view_arr = arr.view()
view_arr[1] = 89
view_arr[-1] = 66
print("Original Array " ,arr)
print("Viewed Array " , view_arr)
```
